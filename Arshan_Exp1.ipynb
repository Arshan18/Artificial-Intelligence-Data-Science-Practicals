## Part 1: Regression 1 (Manual Calculation & Code)

**Given Data:**
*   $X = [2, 3, 2, 3]$
*   $y = [4, 6, 6, 8]$

**Step 1: Find the Means**
*   $\bar{x} = (2 + 3 + 2 + 3) / 4 = 2.5$
*   $\bar{y} = (4 + 6 + 6 + 8) / 4 = 6$

**Step 2: Calculate the Weight (Coefficient / Slope)**
Formula: $w = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sum (x_i - \bar{x})^2}$
*   Numerator: $(-0.5 \times -2) + (0.5 \times 0) + (-0.5 \times 0) + (0.5 \times 2) = 1 + 0 + 0 + 1 = 2$
*   Denominator: $(-0.5)^2 + (0.5)^2 + (-0.5)^2 + (0.5)^2 = 0.25 \times 4 = 1$
*   **Weight ($w$) = 2 / 1 = 2.0**

**Step 3: Calculate the Intercept**
Formula: $b = \bar{y} - w\bar{x}$
*   $b = 6 - (2 \times 2.5) = 6 - 5 = 1.0$
*   **Intercept ($b$) = 1.0**

**Step 4: Predicted Values & Mean Squared Error (MSE)**
*   Predictions ($\hat{y} = wx + b$):
    *   For $x=2$: $(2 \times 2) + 1 = 5$
    *   For $x=3$: $(2 \times 3) + 1 = 7$
*   $\hat{y} = [5, 7, 5, 7]$
*   MSE = $\frac{1}{n} \sum (y_i - \hat{y}_i)^2 = \frac{(-1)^2 + (-1)^2 + (1)^2 + (1)^2}{4} = \frac{4}{4}$ = **1.0**

```python
# Import necessary libraries
import matplotlib.pyplot as plt
import numpy as np
from sklearn import linear_model
from sklearn.metrics import mean_squared_error

# Define the dataset
axis_X = np.array([[2],[3],[2],[3]])
axis_X_train = axis_X
axis_X_test = axis_X

axis_y_train = np.array([4,6,6,8])
axis_y_test = np.array([4,6,6,8])

# Initialize and train the Linear Regression model
model = linear_model.LinearRegression()
model.fit(axis_X_train, axis_y_train)

# Make predictions
axis_y_predicted = model.predict(axis_X_test)

# Print the results to verify manual calculations
print("#### Results ######")
print("Mean squared error is: ", mean_squared_error(axis_y_test, axis_y_predicted))
print("Weights: ", model.coef_)
print("Intercept: ", model.intercept_)

# Plot the actual data points and the regression line
plt.scatter(axis_X_test, axis_y_test, color='blue', label='Actual Data')
plt.plot(axis_X_test, axis_y_predicted, color='red', label='Regression Line')
plt.legend()
plt.show()
```

## Part 2: Regression 2 (Manual Calculation & Code)

**Given Data:**
*   $X = [1, 2, 3, 4, 5, 6, 7]$
*   $y = [9, 8, 10, 12, 11, 13, 14]$

**Step 1: Find the Means**
*   $\bar{x} = 28 / 7 = 4$
*   $\bar{y} = 77 / 7 = 11$

**Step 2: Calculate Weight and Intercept**
*   Using the same formulas as above, the sum of squared differences for $X$ is $28$.
*   The sum of cross-products is $24$.
*   **Weight ($w$)** = $24 / 28 \approx 0.857$
*   **Intercept ($b$)** = $11 - (0.857 \times 4) \approx 7.57$

```python
# Define the dataset
axis_X = np.array([[1],[2],[3],[4],[5],[6],[7]])
axis_X_train = axis_X
axis_X_test = axis_X

axis_y_train = np.array([9,8,10,12,11,13,14])
axis_y_test = np.array([9,8,10,12,11,13,14])

# Initialize and train the Linear Regression model
model = linear_model.LinearRegression()
model.fit(axis_X_train, axis_y_train)

# Make predictions
axis_y_predicted = model.predict(axis_X_test)

# Print the results
print("#### Results ######")
print("Mean squared error is: ", mean_squared_error(axis_y_test, axis_y_predicted))
print("Weights: ", model.coef_)
print("Intercept: ", model.intercept_)

# Plot the actual data points and the regression line
plt.scatter(axis_X_test, axis_y_test, color='blue', label='Actual Data')
plt.plot(axis_X_test, axis_y_predicted, color='red', label='Regression Line')
plt.legend()
plt.show()
```

## Part 3: Diabetes Dataset Implementation

**Step 1: Importing Libraries**
In this step, we import the necessary libraries. `matplotlib.pyplot` is used for data visualization, `numpy` handles array operations, `datasets` allows us to load the built-in diabetes dataset, `linear_model` provides the regression algorithms, and `mean_squared_error` is used to evaluate the model's accuracy.

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model
from sklearn.metrics import mean_squared_error
```

**Step 2: Loading and Formatting the Dataset**
We load the diabetes dataset using `datasets.load_diabetes()`. To perform simple linear regression (one independent variable), we slice the dataset to extract only the third feature (index 2) and add a new axis to convert it into a 2D array, which is required by scikit-learn models.

```python
# Load the built-in diabetes dataset
diabetes = datasets.load_diabetes()

# Print the target variables (the disease progression we want to predict)
print("Target variables length:", len(diabetes.target))

# Extract only one feature (column index 2, which is BMI) for simple linear regression
# np.newaxis increases the dimension of the array to make it a 2D column vector
diabetes_X = diabetes.data[:, np.newaxis, 2]
```

**Step 3: Splitting the Data into Training and Testing Sets**
We need to divide our data so the model can learn from one portion and be tested on unseen data. We take all the data except the last 30 records for training (`[:-30]`), and save the final 30 records specifically for testing (`[-30:]`). We do this for both the features (`X`) and the target variable (`y`).

```python
# Split the features (X) into training and testing sets
diabetes_X_train = diabetes_X[:-30]
diabetes_X_test = diabetes_X[-30:]

# Split the targets (y) into training and testing sets
diabetes_y_train = diabetes.target[:-30]
diabetes_y_test = diabetes.target[-30:]
```

**Step 4: Creating and Training the Model**
We initialize the `LinearRegression()` algorithm and use the `fit()` method to train the model. During training, the model finds the optimal line of best fit by calculating the best weight and intercept for our training data. After training, we use the `predict()` method on our testing data to see how well the model performs.

```python
# Create a linear regression model object
model = linear_model.LinearRegression()

# Train the model using the training sets
model.fit(diabetes_X_train, diabetes_y_train)

# Make predictions using the testing set
diabetes_y_predicted = model.predict(diabetes_X_test)
```

**Step 5: Visualizing the Results**
Finally, we use matplotlib to create a visual representation of our model. The `scatter()` function plots the actual test data points as dots, while the `plot()` function draws the regression line based on the model's predictions.

```python
# Plot the actual test data points
plt.scatter(diabetes_X_test, diabetes_y_test, color='black', label='Actual Test Data')

# Plot the predicted regression line
plt.plot(diabetes_X_test, diabetes_y_predicted, color='blue', linewidth=3, label='Regression Line')

# Display the plot
plt.legend()
plt.show()
```
