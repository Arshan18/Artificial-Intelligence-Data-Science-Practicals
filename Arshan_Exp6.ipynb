```python
import numpy as np

# Max-Min Composition given by Zadeh
def maxMin(x, y):
    z = []
    for x1 in x:
        for y1 in y.T:
            z.append(max(np.minimum(x1, y1)))
    return np.array(z).reshape((x.shape[0], y.shape[1]))

# Max-Product Composition given by Rosenfeld
def maxProduct(x, y):
    z = []
    for x1 in x:
        for y1 in y.T:
            z.append(max(np.multiply(x1, y1)))
    return np.array(z).reshape((x.shape[0], y.shape[1]))

# 3 arrays for the example
r1 = np.array([
    [0.9, 0.2, 0.5],
    [0.4, 0.8, 0.1],
    [0.3, 0.6, 1.0]
])

print("R1:\n", r1)

r2 = np.array([
    [0.7, 0.3, 0.6],
    [0.2, 0.9, 0.4],
    [0.5, 0.1, 0.8]
])

print("R2:\n", r2)

r3 = np.array([
    [1.0, 0.2, 0.4],
    [0.3, 1.0, 0.5],
    [0.6, 0.4, 1.0]
])

print("R3:\n", r3)

print("\n")

print("R1oR2 => Max-Min :\n" + str(maxMin(r1, r2)) + "\n")
print("R1oR2 => Max-Product :\n" + str(maxProduct(r1, r2)) + "\n\n")

print("R1oR3 => Max-Min :\n" + str(maxMin(r1, r3)) + "\n")
print("R1oR3 => Max-Product :\n" + str(maxProduct(r1, r3)) + "\n\n")

print("R1oR2oR3 => Max-Min :\n" + str(maxMin(r1, maxMin(r2, r3))) + "\n")
print("R1oR2oR3 => Max-Product :\n" + str(maxProduct(r1, maxProduct(r2, r3))) + "\n\n")
```

## Code Explanation

This code utilizes the `numpy` library for efficient array operations to demonstrate two ways to combine fuzzy matrices (relations) using composition:

1.  **Max-Min Composition (`Zadeh's Method`)**:
    This is the standard method in fuzzy logic. It evaluates the "weakest link" (minimum) in each possible chain of relationships, and then chooses the "strongest" (maximum) of those weakest links.

2.  **Max-Product Composition (`Rosenfeld's Method`)**:
    Instead of just looking at the minimum, this method multiplies the probabilities (or degrees of membership). This makes it more sensitive to small values, as a small number multiplied by another small number becomes even smaller.

### How the Functions Work
The custom functions, `maxMin` and `maxProduct`, take two matrices (`x` and `y`) as inputs:

*   **The Loops**: Nested `for` loops iterate through each row of the first matrix (`x`) and each column of the second matrix (accessed by transposing `y.T`).

*   **The Math**:
    *   `np.minimum(x1, y1)`: For Max-Min composition, this compares elements to save the smaller numbers.
    *   `np.multiply(x1, y1)`: For Max-Product composition, this multiplies the corresponding elements.

*   **Storing & Reshaping**: The `max(...)` function finds the highest number from the mathematical operation step and appends it to a list called `z`. Finally, `.reshape()` transforms that flat list back into a proper 2D matrix.

### Running the Experiment

1.  We create three 3x3 `numpy` arrays (`r1`, `r2`, and `r3`) with decimal values between 0.0 and 1.0, representing "fuzzy" weights.

2.  We print the initial matrices to observe the starting data.

3.  Finally, we print the results of combining two matrices at a time, as well as chaining all three together (e.g., `R1 o (R2 o R3)`). This vividly demonstrates how Zadeh's and Rosenfeld's methods yield different final answers, highlighting their distinct approaches to fuzzy composition.
