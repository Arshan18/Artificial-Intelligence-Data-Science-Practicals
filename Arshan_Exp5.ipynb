1.   **Classsical set** - A classical set is defined as in which the values are either fully in or fully out , a yes or no .

2.   **Fuzzy set** - A fuzzy set is defined as a set where the members of the set have a degree of membership.


**Operations on Fuzzy set -**

1.  **Union** = It is the maximum of the degree of membership of the two sets
2.  **Intersection** = It is the minimum of the degree of membership of the two sets .
1.  **Complement** = It means negation of the degree of membership i.e 1-degree of membership.
2.  **Substraction** = substraction of two sets a and be is indicated by a-b = a intersection ~b  



1.  **Algebraic Sum** = It is the combination of two fuzzy sets where the membership values are calculated by adding the degree of membership of both sets and subtracting their product.
2.  **Algebraic Product** = It is the multiplication of the degree of membership values of two fuzzy sets.


1.  **Bounded Sum** = It is the addition of the degree of membership values of two fuzzy sets, where the maximum membership value is limited to 1.
2.  **Bounded Product** = It is the subtraction of 1 from the sum of membership values of two fuzzy sets, where negative values are replaced by 0.







Importing the numpy library to perform operaions on the fuzzy set

```python
import numpy as np
```

Declaring the sets A and B i.e giving the degree of membership .

```python

A = np.array([0.2, 0.9, 0.6, 0.4, 0.8])
B = np.array([0.5, 0.4, 0.3, 0.9, 0.7])
```

Union is the maximum of the membership degree of sets A and B

```python
union = np.maximum(A, B)
print("Union:", union)
```

Intersection if the minimum the membership degree of two sets

```python
intersection = np.minimum(A, B)
print("Intersection:", intersection)
```

complement is the negation of a membership degreee of a particular set

```python
complement = 1 - A
print("Complement:", complement)
```

diffrence is substracting membership degree of the two sets , negative values will not be considered

```python
C = 1 - B
print(C)
subtraction = np.minimum(A,C)
print("Subtraction:", subtraction)
```
