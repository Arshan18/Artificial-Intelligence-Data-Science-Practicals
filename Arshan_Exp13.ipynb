```python
# 2. Define the activation function
def step_activation_function(dot_value: int, threshold: float) -> int:
    '''Returns 1 if the dot product meets/exceeds the threshold, else 0'''
    if dot_value >= threshold:
        return 1
    else:
        return 0

# 3. Layer 1: Hidden Neurons Z1 and Z2
weights_z1 = np.array([1, -1])
weights_z2 = np.array([-1, 1])

dot_products_z1 = input_table @ weights_z1
dot_products_z2 = input_table @ weights_z2

T_hidden = 1

# 4. Layer 2: Output Neuron Y
weights_y = np.array([1, 1])
T_output = 1

# 5. Calculate and print results
for i in range(0,4):
    # Get activations for the hidden layer
    act_z1 = step_activation_function(dot_products_z1[i], T_hidden)
    act_z2 = step_activation_function(dot_products_z2[i], T_hidden)

    # Use hidden layer activations as inputs for the final layer
    hidden_inputs = np.array([act_z1, act_z2])
    final_dot = hidden_inputs @ weights_y
    final_activation = step_activation_function(final_dot, T_output)

    print(f'Activation: {final_activation}')
```
